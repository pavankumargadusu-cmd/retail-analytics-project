```python
# ==========================================
# FINAL STEP: REBUILD TABLES & EXPORT
# ==========================================
print("Rebuilding Dimension and Bridge tables with accurate prices...")

# FIX: Fill missing promotions so the filter doesn't return an empty table
if 'Promotion' in df.columns:
    df['Promotion'] = df['Promotion'].fillna('None')

# 1. Rebuild Dimension Table
df_single = df[(df['Product'].apply(len) == 1) & (df['Promotion'] == 'None')].copy()

df_single['Product_Name'] = df_single['Product'].apply(lambda x: x[0])
df_single['Unit_Price'] = df_single['Total_Cost'] / df_single['Total_Items']

dim_products = df_single.groupby('Product_Name')['Unit_Price'].median().reset_index()
dim_products = dim_products.rename(columns={'Unit_Price': 'Estimated_Base_Price'})
dim_products['Product_ID'] = range(1, len(dim_products) + 1)

# 2. Rebuild Bridge Table
bridge_table = df[['Transaction_ID', 'Product']].copy()
bridge_table = bridge_table.explode('Product').rename(columns={'Product': 'Product_Name'})
bridge_table = bridge_table.merge(dim_products[['Product_Name', 'Product_ID']], on='Product_Name', how='left')
bridge_table = bridge_table[['Transaction_ID', 'Product_ID']]

# 3. Rebuild Fact Table
fact_table = df.drop(columns=['Product']).copy()

# 4. Export all to CSV
print("Exporting perfect data to CSV...")
dim_products.to_csv('Dim_Products.csv', index=False)
bridge_table.to_csv('Bridge_Transaction_Products.csv', index=False)
fact_table.to_csv('Fact_Transactions.csv', index=False)

print("Export complete! You are ready for Power BI.")
```
